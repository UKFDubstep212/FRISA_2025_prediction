# Limpieza de Datos


Se eliminaron con dropna(how='all'). Al ser filas enteramente nulas, no hubo riesgo de pérdida de información relevante. El dataset no presentó valores faltantes parciales en ninguna columna.

```python
# Se eliminan filas completamente vacías
df_frisa = df_frisa.dropna(how='all')
df_frisa
```

```python
df_frisa.shape
```

DE Final, DI Final, DE Forja Real 1, DE Forja Real 2, DI Forja Real 1, DI Forja Real 2, Peso Calculado y Peso de Forja Configurado fueron leídas como texto (str) en lugar de numérico. Al inspeccionar sus valores se identificó que usaban coma como separador de miles (ej. 1,841), lo que impidió a pandas reconocerlas como números. Se eliminó la coma con str.replace(',', '') y se convirtieron a float64.

```python
# Se elimina la coma y se convierten a float
cols_coma = ['DE Forja Real 1', 'DE Forja Real 2',
             'DI Forja Real 1', 'DI Forja Real 2',
             'Peso Calculado', 'Peso de Forja Configurado',
             'DE Final', 'DI Final']

for col in cols_coma:
    df_frisa[col] = df_frisa[col].str.replace(',', '').astype(float)
```

Se aplicó el método IQR sobre todas las variables numéricas, excluyendo Orden y N Cliente por ser identificadores sin interpretación matemática. Se encontraron outliers en variables dimensionales y de peso, con los porcentajes más altos en Peso Calculado (10.89%), Peso de Forja Configurado (10.85%) y DE Final (5.38%). Se decidió conservarlos dado que la empresa fabrica anillos de distintas familias geométricas y tamaños, por lo que los valores extremos representan piezas legítimas y no errores de captura.

```python
#Rango Intercuartílico
cols_id = ['Orden', 'N Cliente']

numericas = df_frisa.select_dtypes(include='number').columns.difference(cols_id)

for col in numericas:
    Q1 = df_frisa[col].quantile(0.25)
    Q3 = df_frisa[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df_frisa[(df_frisa[col] < lower) | (df_frisa[col] > upper)]
    print(f"{col}: {len(outliers)} outliers ({len(outliers)/len(df_frisa)*100:.2f}%)")
```

```python
# Boxplots visuales
import matplotlib.pyplot as plt

df_frisa[numericas].plot(kind='box', subplots=True, figsize=(20, 10), layout=(4, -1))
plt.tight_layout()
plt.show()
```

Se identificaron categorías con menos del 1% de frecuencia en Material (56 materiales poco frecuentes), Roladora (P3, P4B, R6) y Familia geométrica (Casquillo 10, Arandela 4). Se conservaron para no perder información. Las variables Motivo y Calidad fueron excluidas de este análisis por requerir tratamiento diferenciado, motivo contiene texto libre descriptivo y Calidad será analizada como variable de interés.

```python
# Detección de categorías raras
excluir = ['Motivo', 'Calidad']

categoricas = df_frisa.select_dtypes(include='object').columns

for col in categoricas:
    if col in excluir:
        continue
    frecuencias = df_frisa[col].value_counts(normalize=True) * 100
    raras = frecuencias[frecuencias < 1]
    if not raras.empty:
        print(col)
        print(raras)
```

Como segundo enfoque de analisis exploratorio la pregunta conductora que elegimos fue: ¿Existe algún patron distintivo en como se comporta el exceso real en comparación con el configurado por familia geométrica?

En primer lugar para este análisis creamos nuevas columnas Diferencia_excesosDE, Diferencia_excesosDI, Diferencia_excesosA.
Esta se calculó como la diferencia entre el promedio de los dos excesos reales medidos y el exceso configurado para cada dimensión.

```python
# Se crea una copia del dataframe para no modificar la estructura original
df_analisis2_frisa = df_frisa
# Creación de los tres nuevos atributos
df_analisis2_frisa['Diferencia_excesosDE'] = (df_analisis2_frisa['DE Exceso Real 1']+ df_analisis2_frisa['DE Exceso Real 2'])/2 - df_analisis2_frisa['Exceso DE configurado']
df_analisis2_frisa['Diferencia_excesosDI'] = (df_analisis2_frisa['DI Exceso Real 1']+ df_analisis2_frisa['DI Exceso Real 2'])/2 - df_analisis2_frisa['Exceso DI Configurado']
df_analisis2_frisa['Diferencia_excesosA'] = (df_analisis2_frisa['Altura Exceso Real 1']+ df_analisis2_frisa['Altura Exceso Real 2'])/2 - df_analisis2_frisa['Exceso Configurado Altura']
df_analisis2_frisa.head()
```

A continuación se agruparon los datos por familia geométrica para analizar datos estadísticos de los tres nuevos atributos

Las cantidades de registros por familia geométrica nos hacen pensar que cualquier conclusion obtenida respecto a arandela 4 o casquillo 10 será poco relevante, debido a que las familias está poco representadas.

Los datos de diametro exterior sugieren que para todas las familias el proceso es estadísticamente inconsistente, ya que se observa una desviación estandar  alta respecto a la media en todos los casos de diametro exterior.

```python
df_por_familia = df_analisis2_frisa.groupby('Familia geométrica')
print('Diferencias de excesos en Diámetro exterior')
df_por_familia['Diferencia_excesosDE'].describe()
```

De igual modo se observa repetidamente una desviación estandar alta respecto a la media en el diametro interior, y es aún más notoria la diferencia que en el diametro exterior. Esto puede deberse a que corregir errores en la cara interior del anillo es un proceso más complejo, por lo que se prefiere dar una mayor holgura para dar espacio en caso de que ocurra un error.

```python
print('Diferencias de excesos en Diámetro interior')
df_por_familia['Diferencia_excesosDI'].describe()
```

En la altura se aprecia que la media es consistentemente negativa. Esto puede significar que el algoritmo consistentemente asigna altura de menos a las piezas. Nuevamente se observa una desviación estandar atípicamente alta, de nuevo sugiriendo un proceso inconsistente

```python
print('Diferencias de excesos en Altura')
df_por_familia['Diferencia_excesosA'].describe()
```

Para visualizar mejor el comportamiento de estos nuevos atributos se generaron los siguientes boxplot.

Como se pronosticó, en todas las familias se observa una variación muy alta. Además, se observa que el exceso real que conserva el diametro exterior es consistentemente más alto que el configurado, lo cual sugiere que el algoritmo asigna exceso de más a esa dimensión.

```python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn import preprocessing

plt.figure(figsize=(12, 6))
sns.boxplot(x=df_analisis2_frisa['Familia geométrica'], y=df_analisis2_frisa['Diferencia_excesosDE'])
plt.title('Distribución de diferencia de excesos en diametro exterior por Familia Geométrica')
plt.xlabel('Familia Geométrica')
plt.ylabel('Diferencia de excesos en diametro exterior')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

En la dimensión de diametro interior la media es más cercana a cero, lo cual significa que en promedio el algoritmo asigna correctamente el exceso, pero la variabilidad es también muy alta.

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x=df_analisis2_frisa['Familia geométrica'], y=df_analisis2_frisa['Diferencia_excesosDI'])
plt.title('Distribución de diferencia de excesos en diametro interior por Familia Geométrica')
plt.xlabel('Familia Geométrica')
plt.ylabel('Diferencia de excesos en diametro interior')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
